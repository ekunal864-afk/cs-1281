```python
logs = [
    "P001, 25.4, 3.5",
    "P002,30.0, 2.1",
    "P003,  22.7 , 4.0",
    "P004,27.5, 1.8",
    "P005,  31.2,3.2",
    "P006,29.8 , 2.5",
    "P007, 24.6, 3.9",
    "P008,  35.0,1.5",
    "P009,26.3 , 2.8",
    "P010, 28.9, 3.1"
]

diameters = []

for line in logs:
    parts = line.split(",")
    dia = float(parts[1].strip())
    diameters.append(dia)

count = len(diameters)
average = sum(diameters) / count
minimum = min(diameters)
maximum = max(diameters)

print("Count:", count)
print("Average:", average)
print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
logs = [
    "P001, 25.4, 3.5",
    "P002,30.0, 2.1",
    "P003,  22.7 , 4.0",
    "P004,27.5, 1.8",
    "P005,  31.2,3.2",
    "P006,29.8 , 2.5",
    "P007, 24.6, 3.9",
    "P008,  35.0,1.5",
    "P009,26.3 , 2.8",
    "P010, 28.9, 3.1"
]

accepted = []
rework = []
scrap = []
scrapped_parts = []

for line in logs:
    parts = line.split(",")

    part_code = parts[0].strip()
    dia = float(parts[1].strip())

    if 25.0 <= dia <= 30.0:
        accepted.append(dia)

    elif 20.0 <= dia < 25.0 or 30.0 < dia <= 32.0:
        rework.append(dia)

    else:
        scrap.append(dia)
        scrapped_parts.append(part_code)

print("Accepted count:", len(accepted))
print("Rework count:", len(rework))
print("Scrap count:", len(scrap))

if len(accepted) > 0:
    print("Accepted average:", sum(accepted) / len(accepted))
else:
    print("Accepted average: No accepted parts")

print("Scrapped part codes:", scrapped_parts)
```

```python
day_logs = [
    "P001,25.4,3.5",
    "P002,30.0,2.1",
    "P003,22.7,4.0",
    "P004,27.5,1.8",
    "P005,35.0,3.2"
]

night_logs = [
    "P003,35.5,2.0",
    "P006,29.8,2.5",
    "P007,24.6,3.9",
    "P005,36.0,1.5",
    "P008,28.9,3.1"
]

def process_shift(logs):
    accepted = []
    scrap_parts = []

    for line in logs:
        parts = line.split(",")

        part_code = parts[0].strip()
        dia = float(parts[1].strip())

        if 25.0 <= dia <= 30.0:
            accepted.append(part_code)
        else:
            scrap_parts.append(part_code)

    accept_rate = len(accepted) / len(logs) * 100

    return accept_rate, scrap_parts


day_rate, day_scrap = process_shift(day_logs)
night_rate, night_scrap = process_shift(night_logs)

print("Day accept rate:", day_rate, "%")
print("Night accept rate:", night_rate, "%")

# Convert scrap lists into sets
day_scrap_set = set(day_scrap)
night_scrap_set = set(night_scrap)

# Parts scrapped in BOTH shifts
both_scrap = day_scrap_set & night_scrap_set

# Parts appearing only in day scrap
day_only = day_scrap_set - night_scrap_set

# Parts appearing only in night scrap
night_only = night_scrap_set - day_scrap_set

print("Scrapped in both shifts:", both_scrap)
print("Scrapped only during day:", day_only)
print("Scrapped only during night:", night_only)
```
