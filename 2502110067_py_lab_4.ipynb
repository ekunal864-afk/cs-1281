```python
mode = input("Select mode (halt/skip): ").strip().lower()

inspected = 0

for i in range(100):
    dia = float(input("Measured diameter: "))
    inspected += 1

    if dia > 25.05:
        print("REWORK")   # oversize — metal can come off

    elif dia < 24.95:
        print("SCRAP")    # undersize — metal never goes back

        if mode == "halt":
            print(f"Batch halted. Inspected: {inspected}")
            break
        elif mode == "skip":
            continue

    else:
        print("ACCEPT")

if mode == "skip":
    print(f"Batch complete. Inspected: {inspected}")
```

```python
consecutive_accepts = 0
attempts = 0
MAX_ATTEMPTS = 5

while consecutive_accepts < 5 and attempts < MAX_ATTEMPTS:
    dia = float(input("Measured diameter: "))
    attempts += 1

    if dia > 25.05:
        print("REWORK")
        consecutive_accepts = 0

    elif dia < 24.95:
        print("SCRAP")
        consecutive_accepts = 0

    else:
        print("ACCEPT")
        consecutive_accepts += 1

if consecutive_accepts == 5:
    print(f"Five consecutive ACCEPTs achieved in {attempts} parts.")
else:
    print(f"Safety cap reached after {attempts} attempts.")
    print("Five consecutive ACCEPTs were not achieved.")
```
